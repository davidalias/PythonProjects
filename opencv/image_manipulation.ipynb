```python
import cv2
import random
```

```python
img = cv2.imread("assets/image.JPG", -1)
```

```python
print(img) #loads the image into a numpy array
```

```python
print(img.shape)
```

```python
print(img[0][45:400])
```

```python
#modifies the top 100 rows of the image by replacing every pixel with a random color
for i in range(100):
    for j in range(img.shape[1]):
        img[i][j] = [random.randint(0,255), random.randint(0,255), random.randint(0,255)]
```

```python
cv2.imshow("Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
