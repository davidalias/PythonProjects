```python
import cv2
import numpy as np
```

```python
cap = cv2.VideoCapture(0)
```

```python
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")
```

```python
while True:
    ret, frame = cap.read()

    grey = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY) #to convert image to greyscale to perform classification
    faces = face_cascade.detectMultiScale(grey, 1.3, 5) #return location of all of face positions. 1.3 -> scale down the image by 30% (recommended 1.05). minNeighbors = 5

    for (x,y,w,h) in faces:
        #face detection within the region of interest of frame
        cv2.rectangle(frame, (x,y), (x+w, y+h), (255,0,0), 5) #for detecting the face position in RGB input frame
        roi_grey = grey[y:y+h,x:x+w] #for transferring grey scale image to eye cascade classifier
        roi_color = frame[y:y+h, x:x+w] #for transferring face ROI RGB frame for eye detection

        #eye detection within the region of interest of face
        eyes = eye_cascade.detectMultiScale(roi_grey, 1.3, 5)
        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex,ey), (ex+ew, ey+eh), (0,255,0), 5)#for detecting the eye position within the RGB input frame

    cv2.imshow("frame", frame)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
