```python
import cv2
import numpy as np
```

```python
img = cv2.imread("assets/image.JPG")
img = cv2.resize(img, (0,0), fx = 1.5, fy=1.5)
```

```python
#convert to greyscale for better algo work
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
corners = cv2.goodFeaturesToTrack(gray, 6, 0.01, 5)#6 best corners will be returned with 0.01 confidence with minimum euclidean distance of 5
print(corners)
```

```python
corners = np.int0(corners)
corners
```

```python
for corner in corners:
    x,y = corner.ravel() #[[x,y]] -> [x,y]
    cv2.circle(img, (x,y), 5, (255,0,0), -1)
```

```python
#To draw lines between corner points
for i in range(len(corners)):
    for j in range(i+1, len(corners)):
        corner1 = tuple(corners[i][0])
        corner2 = tuple(corners[j][0])
        color = tuple(map(lambda x: int(x), np.random.randint(0,255, size =3)))#random colour generation
        cv2.line(img, corner1, corner2, color, 1)
```

```python
cv2.imshow("Frame", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
